```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
os.listdir()
```

```python
df=pd.read_excel("dataset 2.1.csv.xlsx")
df.head()
```

```python
df.tail()
```

```python
# Shape of data
print("Dataset Shape:", df.shape)

# Column info
df.info()
```

```python
# Missing values
df.isnull().sum()
```

```python
# Duplicate rows
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

```python
df.describe()
```

```python
# Number of unique images
num_images = df['frame'].nunique()
print("Number of unique images:", num_images)

# Number of classes
num_classes = df['class_id'].nunique()
print("Number of unique classes:", num_classes)

# Class labels
sorted(df['class_id'].unique())
```

```python
class_counts = df['class_id'].value_counts().sort_index()
class_counts
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=class_counts.index, y=class_counts.values)
plt.title("Class Distribution")
plt.xlabel("Class ID")
plt.ylabel("Count")
plt.show()
```

```python
df['bbox_width'] = df['xmax'] - df['xmin']
df['bbox_height'] = df['ymax'] - df['ymin']
df['bbox_area'] = df['bbox_width'] * df['bbox_height']

df[['bbox_width', 'bbox_height', 'bbox_area']].describe()
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
sns.histplot(df['bbox_width'], bins=50, kde=True)
plt.title("Bounding Box Width Distribution")

plt.subplot(1,2,2)
sns.histplot(df['bbox_height'], bins=50, kde=True)
plt.title("Bounding Box Height Distribution")

plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.histplot(df['bbox_area'], bins=50, kde=True)
plt.title("Bounding Box Area Distribution")
plt.show()
```

```python
boxes_per_image = df.groupby('frame').size()

boxes_per_image.describe()
```

```python
plt.figure(figsize=(7,5))
sns.histplot(boxes_per_image, bins=30)
plt.title("Bounding Boxes per Image")
plt.xlabel("Boxes per image")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='class_id', y='bbox_area', data=df)
plt.yscale("log")
plt.title("Bounding Box Area by Class (log scale)")
plt.show()
```

```python
# Very large boxes
df.sort_values("bbox_area", ascending=False).head(10)
```

```python
plt.figure(figsize=(6,5))
sns.heatmap(
    df[['xmin','ymin','xmax','ymax','bbox_width','bbox_height','bbox_area']].corr(),
    annot=True, cmap="coolwarm"
)
plt.title("Correlation Matrix")
plt.show()
```

```python
explode = [0.05] * len(class_counts)

plt.figure(figsize=(8,8))
plt.pie(
    class_counts.values,
    labels=class_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    explode=explode,
    shadow=True
)
plt.title("Class Distribution (Exploded Pie Chart)")
plt.axis('equal')
plt.show()
```

Exploratory Data Analysis (EDA) – Conclusion

The exploratory data analysis was conducted to understand the structure, quality, and statistical characteristics of the object detection dataset. The dataset contains a large number of bounding box annotations distributed across multiple images and object classes, making it suitable for training and evaluating object detection models.

Key Findings

The dataset consists of 165,105 object annotations across a substantial number of unique images, indicating a rich and diverse dataset.

Multiple object classes are present, however, the class distribution analysis reveals a noticeable class imbalance, where certain classes dominate the dataset while others are underrepresented.

Bounding box analysis shows a wide variation in object sizes, with bounding box widths, heights, and areas spanning multiple scales. This suggests the presence of both small and large objects within images.

The distribution of bounding boxes per image indicates that most images contain a limited number of objects, while a smaller subset of images contains a higher object density.

Correlation analysis confirms expected relationships between bounding box dimensions and area, validating the consistency of annotation data.

A small number of extreme bounding box sizes were observed, which may represent valid large objects or potential annotation outliers.

Data Quality Assessment

The dataset contains no critical missing values or structural inconsistencies, indicating good annotation quality.

Duplicate entries are minimal and do not significantly impact overall data integrity.

Implications for Model Development

Due to class imbalance, techniques such as data augmentation, class weighting, or focal loss may be required to improve model performance on minority classes.

The large variation in object sizes highlights the need for multi-scale feature extraction, making models like YOLO, SSD, or Faster R-CNN suitable choices.

Outlier bounding boxes should be reviewed during preprocessing to ensure they do not negatively impact training stability.

Final Assessment

Overall, the dataset is well-structured, sufficiently large, and appropriate for object detection tasks. With proper preprocessing and imbalance handling, it is expected to support robust model training and generalization.
